```python
from sklearn.model_selection import TimeSeriesSplit
#need this so we can have past predicft future not the other way around
```

```python
from sklearn.linear_model import RidgeClassifier
```

```python
rr = RidgeClassifier(alpha=1)
split = TimeSeriesSplit(n_splits=3)
sfs=SequentialFeatureSelector(rr, n_features_to_select=30,direction="forward", cv=split)
```

```python
# we need to scale the data for ridge regression... we are scaling are columns to fall between 0 and 1 
removed_columns = ["season","date","won","target","team","team_opp"]
#we are not scaling these columsn above
selected_columns = df.columns[~df.columns.isin(removed_columns)]
#we are scaling these collumns 
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[selected_columns] = scaler.fit_transform(df[selected_columns])
```

```python
sfs.fit(df[selected_columns],df["target"])
```

```python
predictors = list(selected_columns[sfs.get_support()])
predictors
```

```python
def backtest(data, model, predictors, start=2, step=1):
    #split data up by season and use past season to predicft future seasons
    all_predictions = []
    seasons = sorted(data["season"].unique())

    for i in range(start, len(seasons), step):
        season = seasons[i]
        train = data[data["season"] < season]
        test = data[data["season"] == season]

        model.fit(train[predictors], train["target"])
        #training the model 

        preds = model.predict(test[predictors])
        #this is prediction the test set 
        preds = pd.Series(preds, index=test.index)
        combined = pd.concat([test["target"], preds], axis = 1)
        combined.columns = ["actual", "prediction"]

        all_predictions.append(combined)
    return pd.concat(all_predictions)
```

```python
predictions = backtest(df, rr, predictors)
```

```python
predictions
```

```python
from sklearn.metrics import accuracy_score 
test = predictions[predictions["actual"] != 2]
accuracy_score(test["actual"], test["prediction"])
#calculating the accurary of are models
```
